```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
with open("calendar.html", "r", encoding="utf-8") as f:
      soup = BeautifulSoup(f, "html.parser")

container = soup.find("div", class_="body-paragraph")
tbody = container.find("tbody")
```

```python
date, day, note = [], [], []

for row in tbody.find_all("tr"):
    cells = row.find_all("td")
    if len(cells) == 3:
        date.append(cells[0].text.strip())
        day.append(cells[1].text.strip())
        note.append(cells[2].get_text(separator=" ", strip=True))

data_pds = pd.DataFrame({"Date" : date, "dow" : day, "text" : note})
```

```python
def parse_date(date_str):
      start_part = date_str.split("-")[0].strip()
      if "2021" not in start_part and "2022" not in start_part:
          start_part = f"{start_part} 2021"
      return pd.to_datetime(start_part)


data_pds["Date"] = data_pds["Date"].apply(parse_date)
data_pds = data_pds.set_index("Date")

print(data_pds)
```
